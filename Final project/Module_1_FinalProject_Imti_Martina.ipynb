## Project Title:
Associations Between ABeta42 Levels, Dementia Status, Sex, and Age at Death



## Project Goal:

Amyloid-beta plaques are a major pathological feature of Alzheimer’s disease, and understanding the factors associated with variation in ABeta42 concentrations can help characterize differences related to the disease. This project seeks to determine the relevance, if any, of ABeta42 in the context of factors associated with Alzheimer’s disease by attempting to answer the following three questions:

1. Is there a significant difference in mean ABeta42 concentrations between male and female donors with dementia?

2. Does the mean ABeta42 concentration differ among male and female donors with and without dementia?

3. Is there a relationship between age at death and ABeta42 concentration?

To answer these questions, an independent t-test will be used to compare male and female donors with dementia, a one-way ANOVA will be used to compare the four sex and dementia-status groups, and a linear regression model will be used to evaluate the relationship between age at death and ABeta42 concentration.

* Diagnosis:
    
    Diagnosis of Alzheimer’s is a multi-step process in which doctors use several tools, including medical history, cognitive testing, brain imaging, and biomarker testing. Biomarkers such as amyloid-beta and tau can be measured by collecting cerebrospinal fluid (CSF) through a spinal tap. In addition, doctors may perform brain scans such as CT, MRI, and PET scans to support the diagnosis and help rule out other causes of cognitive impairment. These tests may also be repeated over time to monitor changes in cognitive function. If a person is diagnosed with Alzheimer’s, they may be referred to specialists such as geriatricians, geriatric psychiatrists, neurologists, or neuropsychologists.

Lastly, in the severe stage, patients may lose the ability to communicate, move independently, and perform activities necessary to care for themselves. At this stage, they often require around-the-clock assistance with their daily routine.

* Biological mechanisms (anatomy, organ physiology, cell & molecular physiology):
    
    Alzheimer’s disease primarily affects the hippocampus region of the brain, which is important for the formation and storage of memories. As the disease progresses, it affects other parts of the cerebral cortex, and the cumulative loss of neurons leads to brain atrophy.

These changes interfere with communication between neurons and can eventually cause them to die. This loss of neurons contributes to the decline in cognitive function.

## Data Analysis: 

The data were analyzed using Python. Each row of the CSV file was converted into a Patient object containing the donor’s demographic, cognitive, genetic, and protein data. The analysis examined ABeta42 concentrations using an independent t-test, a one-way ANOVA, and linear regression. A significance level of p < 0.05 was used for the statistical tests.

Citation for the source of csv: 

Gabitto, M. I., Travaglini, K. J., Rachleff, V. M., Kaplan, E. S., Long, B., Ariza, J., Ding, Y., Mahoney, J. T., Dee, N., Goldy, J., Melief, E. J., Agrawal, A., Kana, O., Zhen, X., Barlow, S.
```python
from patient import *  # imports the Patient class from patient.py
import matplotlib.pyplot as plt  # creates graphs
import numpy as np  # creates error bars
from scipy import stats  # performs statistical tests
import statistics  # calculates means and standard deviations
import pandas as pd  # reads CSV files
from sklearn.linear_model import LinearRegression  # creates regression model; download sci-kit learn if you don't have it

# Prevent patients from being duplicated if this cell is run again
Patient.all_patients.clear()

# Create one Patient object for each donor in the CSV file
Patient.instantiate_from_csv(
    r"Metadata and Protein Data for Module 1.csv" # file location
)

# Confirm that all 84 patients were loaded correctly
print("Number of patients:", len(Patient.all_patients))
print("First patient:", Patient.all_patients[0])

# Sort patients by ABeta42 level from lowest to highest
Patient.all_patients.sort(
    key=Patient.get_abeta42,
    reverse=False
)
```

The donors were then divided into four groups according to their sex and cognitive status. The categories were females with dementia, males with dementia, females without dementia, and males without dementia. The ABeta42 measurements were extracted from each group so that their means and standard deviations could be calculated.

```python
# Female donors with dementia
female_dementia_patients = Patient.filter(
    Patient.all_patients,
    sex="Female",
    cognitive_status="Dementia"
)

# Female donors without dementia
female_no_dementia_patients = Patient.filter(
    Patient.all_patients,
    sex="Female",
    cognitive_status="No dementia"
)

# Male donors without dementia
male_no_dementia_patients = Patient.filter(
    Patient.all_patients,
    sex="Male",
    cognitive_status="No dementia"
)

# Male donors with dementia
male_dementia_patients = Patient.filter(
    Patient.all_patients,
    sex="Male",
    cognitive_status="Dementia"
)

# Empty lists to hold the ABeta42 measurements
female_abeta42 = []
male_abeta42 = []
female_no_dementia_abeta42 = []
male_no_dementia_abeta42 = []

# Add each donor's ABeta42 measurement to the appropriate list
for patient in female_no_dementia_patients:
    female_no_dementia_abeta42.append(patient.abeta42)

for patient in male_no_dementia_patients:
    male_no_dementia_abeta42.append(patient.abeta42)

for patient in female_dementia_patients:
    female_abeta42.append(patient.abeta42)

for patient in male_dementia_patients:
    male_abeta42.append(patient.abeta42)

# Calculate the group means
female_abeta42_mean = statistics.mean(female_abeta42)
male_abeta42_mean = statistics.mean(male_abeta42)

female_no_dementia_abeta42_mean = statistics.mean(
    female_no_dementia_abeta42
)
male_no_dementia_abeta42_mean = statistics.mean(
    male_no_dementia_abeta42
)

# Calculate the group standard deviations
female_no_dementia_abeta42_stdev = statistics.stdev(
    female_no_dementia_abeta42
)
male_no_dementia_abeta42_stdev = statistics.stdev(
    male_no_dementia_abeta42
)

female_abeta42_stdev = statistics.stdev(female_abeta42)
male_abeta42_stdev = statistics.stdev(male_abeta42)

print("Female dementia ABeta42 mean:", female_abeta42_mean)
print(
    "Female dementia ABeta42 standard deviation:",
    female_abeta42_stdev
)

print("Male dementia ABeta42 mean:", male_abeta42_mean)
print(
    "Male dementia ABeta42 standard deviation:",
    male_abeta42_stdev
)
```

Then, an independent t-test was used to compare mean ABeta42 concentrations between female and male donors with dementia. This test was deemed appropriate because ABeta42 is a numerical variable and the analysis compared two separate, independent groups. The null hypothesis was that female and male donors with dementia had equal mean ABeta42 concentrations.

```python
# Perform an independent t-test
t_stat, p_val = stats.ttest_ind(
    female_abeta42,
    male_abeta42
)

print("t-statistic:", t_stat)
print("T-test p-value:", p_val)
```

A one-way ANOVA was used to compare mean ABeta42 concentrations among the four groups defined by sex and dementia status. ANOVA was deemed appropriate because the analysis compared the mean of one numerical variable across more than two independent groups. The null hypothesis was that all four groups had equal mean ABeta42 concentrations.

```python
# Perform a one-way ANOVA across the four groups
f_stat, p_value = stats.f_oneway(
    female_no_dementia_abeta42,
    male_no_dementia_abeta42,
    female_abeta42,
    male_abeta42
)

print("F-statistic:", f_stat)
print("ANOVA p-value:", p_value)
```

The first bar graph compares the mean ABeta42 concentrations of female and male donors with dementia. The error bars represent one standard deviation, and the t-statistic and p-value are displayed on the graph.

```python
patient_sex_groups = [
    "Female Patients",
    "Male Patients"
]

mean_abeta42 = [
    female_abeta42_mean,
    male_abeta42_mean
]

stdev_abeta42 = [
    female_abeta42_stdev,
    male_abeta42_stdev
]

# Create upward error bars representing one standard deviation
yerr = [
    np.zeros(len(mean_abeta42)),
    stdev_abeta42
]

plt.bar(
    patient_sex_groups,
    mean_abeta42,
    yerr=yerr,
    capsize=10,
    color=["purple", "blue"]
)

plt.title("Average ABeta42 Levels in Patients with Dementia")
plt.xlabel("Sex")
plt.ylabel("Average ABeta42 Level (pg/ug)")

y_max = max(mean_abeta42) + max(stdev_abeta42) * 0.4

plt.text(
    0.5,
    y_max,
    f"t = {t_stat:.2f}\np = {p_val:.3e}",
    ha="center",
    va="bottom"
)

plt.show()
```

The second bar graph compares the mean ABeta42 concentrations among all four sex and dementia-status groups. The error bars represent one standard deviation, and the overall ANOVA p-value is displayed on the graph.

```python
anova_groups = [
    "No Dementia Female",
    "No Dementia Male",
    "Dementia Female",
    "Dementia Male"
]

anova_means = [
    female_no_dementia_abeta42_mean,
    male_no_dementia_abeta42_mean,
    female_abeta42_mean,
    male_abeta42_mean
]

anova_stdev = [
    female_no_dementia_abeta42_stdev,
    male_no_dementia_abeta42_stdev,
    female_abeta42_stdev,
    male_abeta42_stdev
]

anova_yerr = [
    np.zeros(len(anova_means)),
    anova_stdev
]

plt.bar(
    anova_groups,
    anova_means,
    yerr=anova_yerr,
    capsize=10,
    color=["red", "blue", "pink", "skyblue"]
)

plt.title("ABeta42 Levels")
plt.xlabel("Sex and Dementia Status")
plt.ylabel("Average ABeta42 Level (pg/ug)")

plt.text(
    1.5,
    380,
    f"One-Way ANOVA: p = {p_value:.3f}",
    ha="right",
    va="top",
    fontsize=12
)

plt.show()
```

Lastly, a scatterplot and linear regression were used to examine the relationship between age at death and ABeta42 concentration because both of the variables are numerical. Age at death was the independent variable, and ABeta42 was the dependent variable. The regression slope represents the predicted change in ABeta42 for each additional year of age, while R² represents the proportion of variation in ABeta42 explained by its linear relationship with age.

```python
patient_ages = []
patient_abeta42 = []

# Retrieve each donor's age at death
for patient in Patient.all_patients:
    patient_ages.append(patient.age_at_death)

# Retrieve each donor's ABeta42 measurement
for patient in Patient.all_patients:
    patient_abeta42.append(patient.abeta42)

# Prepare the independent and dependent variables
X = np.array(patient_ages).reshape(-1, 1)
y = np.array(patient_abeta42)

# Create and fit the regression model
model = LinearRegression()
model.fit(X, y)

# Obtain the regression values
slope = model.coef_[0]
intercept = model.intercept_
r2 = model.score(X, y)

equation = (
    f"y = {slope:.2f}x + {intercept:.2f}"
    f"\nR² = {r2:.2f}"
)

print("Slope:", slope)
print("Intercept:", intercept)
print("R-squared:", r2)

# Create the scatterplot and regression line
plt.text(
    X.min(),
    y.max(),
    equation,
    color="red",
    fontsize=12,
    verticalalignment="top"
)

plt.scatter(X, y, color="green")
plt.plot(X, model.predict(X), color="red")

plt.xlabel("Age at Death (years)")
plt.ylabel("ABeta42 Level (pg/ug)")
plt.title("ABeta42 Level vs. Age at Death")

plt.show()
```

## Verify and validate your analysis: 

To verify the analysis, we checked that the calculated means, standard deviations, statistical results, and regression values matched the values displayed in the figures.

The independent t-test produced t = 0.750 and p = 0.458. Therefore, the analysis did not yield any statistically significant difference in mean ABeta42 concentration between female and male donors with dementia. The one-way ANOVA produced F = 1.402 and p = 0.248, which indidcates that the analysis did not find a statistically significant difference among the four sex and cognitive-status groups. The regression slope was -2.75 pg/ug per year, with R² = 0.02. 

The results were compared with published Alzheimer’s disease research. Serrano-Pozo et al. (2011) identified amyloid plaques as a major feature of Alzheimer’s disease, but that amyloid accumulation can begin before cognitive symptoms appear. This supports the overlap in our dataset in ABeta42 concentrations between donors with and without dementia. A paper published by Oveisgharan et al. (2018) found that women had greater overall Alzheimer’s development than men, but the sex difference in amyloid-beta load was only borderline significant. Our female donors with dementia had a higher mean ABeta42 concentration than male donors with dementia, but our t-test also did not show a statistically significant sex difference. Overall, our results are generally consistent with published evidence that sex differences in amyloid development may be present but are not necessarily statistically significant. 

## Conclusions and Ethical Implications: 

For differences in ABeta42 levels across sex, our analysis found no statistically significant difference in ABeta42 levels between male and female patients. Furthermore, when examining the donors based on both dementia status and sex groups, donors with dementia had higher mean ABeta42 concentrations than donors without dementia; however, the analysis did not provide sufficient evidence to conclude that this difference was statistically significant. Lastly, for age at death, the low R^2 value indicates that age explained only about 1.9% of the variation in ABeta42 concentration in the linear regression model.

The inability of age, sex, and dementia status to explain the variability in ABeta42 concentration does not mean that these factors have no relationship with Alzheimer’s disease. Rather, it suggests that the results are specific to this sample and may be influenced by factors such as unequal sample sizes and outliers. Because of these statistical limitations, using the findings from this analysis to make broader clinical conclusions should be approached with caution, as doing so could lead to inaccurate interpretations.

## Limitations and Future Work: 


Our analysis did not find statistically significant differences in ABeta42 concentration based on sex or dementia status. Female donors with dementia did have a higher mean ABeta42 concentration (116.09 pg/ug) than male donors with dementia (63.15 pg/ug), but the independent t-test was not statistically significant (t = 0.750, p = 0.458). Thus, the null hypothesis could not be rejected, and this dataset does not provide sufficient evidence that female and male donors with dementia have different mean ABeta42 concentrations.

The female and male dementia groups also had higher mean ABeta42 concentrations than their corresponding no-dementia groups. However, the one-way ANOVA was not statistically significant (F = 1.402, p = 0.248), meaning that the dataset does not provide sufficient evidence that mean ABeta42 differs among the four sex and cognitive-status groups. The regression slope was -2.75 pg/ug per year, but the R² value was only 0.02. This was interpreted to mean that age at death explained approximately 1.9% of the variation in ABeta42, and was not a useful predictor of ABeta42 concentration in this dataset.

That being said, several limitations should be considered. The groups were relatively small and unequal in size, with non-equal numbers of male and female samples. Additionally, the ABeta42 measurements showed substantial variability, with multiple outliers that were not excluded during statistical analysis. One female donor with dementia had an ABeta42 concentration of approximately 1,412.57 pg/ug, which was much higher than the other measurements and strongly affected the group mean and standard deviation. The observation was retained because the dataset did not identify it as an error and to yield the most inclusive results. 

These findings should not be used to conclude that sex, age, or dementia status has no biological relationship with ABeta42. The results only show that no statistically significant relationships were detected in this sample. It would also be unethical to use these group-level findings to predict an individual’s dementia status or Alzheimer’s disease risk. Doing so could contribute to inaccurate diagnoses or stereotypes about women, men, or older adults.

It seems reasonable to conclude that from the analysis done, ABeta42 should not be used alone to make clinical conclusions about an individual. We recommend that future evaluations use significantly larger and more gender-balanced donor groups and consider implementing information such as those given in the dataset including tau concentrations, APOE genotype, neuropathological findings, and clinical history.
